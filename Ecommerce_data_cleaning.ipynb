## Data Cleaning

```python
# Remove duplicates
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# Replace common invalid values with NaN
df.replace(['N/A', 'NULL', ''], np.nan, inplace=True)
```

```python
# Remove extra spaces
df = df.apply(lambda x : x.str.strip() if x.dtype == 'object' else x).copy()
```

```python
# Standardize text format
# Convert text into Proper case

cols = ['Customer_Name', 'City', 'State']

for i in cols:
    df[i] = df[i].str.title()
```

```python
# Validate 'Email'
# Keep only valid email records

df = df[df['Email'].str.contains('@', na=False)]
```

```python
# Remove invalid 'Qty' (negative or zero)

df = df[df['Qty']>0]
```

```python
# Convert date columns

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce', dayfirst=True)
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'], errors='coerce', dayfirst=True)
```

```python
# Convert numeric columns

cols = ['Qty', 'Unit_Price', 'Discount']

for i in cols:
    df[i] = pd.to_numeric(df[i], errors='coerce')
```

```python
# Filling Missing Values

df['Discount'] = df['Discount'].fillna(0)
df['Phone'] = df['Phone'].fillna('Unknown')
df['Delivery_Date'] = df['Delivery_Date'].fillna(0)
```

## Feature Engineering

```python
# Sales

df['Sales'] = df['Qty']*df['Unit_Price']
```

```python
df.head()
```

```python
# Net Amount

df['Net_Amount'] = df['Sales'] - (df['Sales']*df['Discount'])
```

```python
df.head()
```

```python
# Profit (Assuming 20%)

df['Profit'] = df['Net_Amount']*0.20
```

```python
df.head()
```

```python
# Order Month
df['Month'] = df['Order_Date'].dt.month_name()
```

```python
# Year
df['Year'] = df['Order_Date'].dt.year
```

```python
# Weekday
df['Weekday'] = df['Order_Date'].dt.day_name()
```
